What else can your Class be programmed to do?

*(Oh the `IPython.display` library has a function called `clear_output` which will clear the old output so that you can overlay the new output and give the impression of movement)*

````python
class Room:
    def __init__(self, width, height):
        # gives room dimentions
        self.width = width
        self.height = height
        # player position
        self.x = 0
        self.y = 0

    def draw(self):
        print() # gives a empty space to look nicer
        print("#" * (self.width + 2)) # gives a boarder
        for row in range(self.height): # for each roq
            line = "#" # starts each line with left boarder
            for col in range(self.width): # for each column
                if col == self.x and row == self.y: # compares player positon and current positon
                    line += "@" # draws player
                else:
                    line += "." # empty space
            line += "#" # right boarder
            print(line)
        print("#" * (self.width + 2)) # bottom boarder
        print("a=left  s=right  q=up  z=down  x=quit") # display controls 
# all movement 
    def left(self):
        if self.x > 0:
            self.x -= 1

    def right(self):
        if self.x < self.width - 1:
            self.x += 1

    def up(self):
        if self.y > 0:
            self.y -= 1

    def down(self):
        if self.y < self.height - 1:
            self.y += 1


myRoom = Room(4, 4) # create room
myRoom.draw() # draw room

# main loop
while True:
    # get input
    s = input("Move: ")
    # deals with movement inpuy
    if s == 'a': myRoom.left()
    if s == 's': myRoom.right()
    if s == 'q': myRoom.up()
    if s == 'z': myRoom.down()
    if s == 'x':
        print("all done")
        break
    # redraws the room with updated player position
    myRoom.draw()
````
