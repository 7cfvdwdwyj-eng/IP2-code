````python
# Write your set-up and calculations for question 1
theta_deg = 25
theta_rad = np.radians(theta_deg)
v0 = 7.5
g = 9.81

t_flight = (2 * v0 * np.sin(theta_rad)) / g
print(f"total flight time = {t_flight:.3f}s")

t = np.linspace(0, t_flight, 500)
#np.linspace makes the points evently spread throughout
````

````python
# Write your calculations for question 2 here
# creates an array with 500 values of v_x allowing for it to be plotted against 500 t values
v_x = np.full_like(t, v0 * np.cos(theta_rad)) 
v_y = v0 * np.sin(theta_rad) - g * t
v_mag = np.sqrt(v_x**2 + v_y**2)
p_x = v0 * np.cos(theta_rad) * t 
p_y = v0 * np.sin(theta_rad) * t - 0.5 * g * t**2
````

````python
import numpy as np
import matplotlib.pyplot as plt
# Make your plots here
plt.figure(figsize=[10,10])

# v componets 
plt.subplot(2, 2, 1)
plt.plot(t, v_x, label="Horizontal velocity $v_x$")
plt.plot(t, v_y, label="Vertical velocity $v_y$")
plt.xlabel("Time (s)")
plt.ylabel("Velocity (m/s)")
plt.title("Velocity components")
plt.legend()
plt.grid(True)

#speed
plt.subplot(2, 2, 2)
plt.plot(t, v_mag, label="Speed")
plt.xlabel("Time (s)")
plt.ylabel("Speed (m/s)")
plt.title("Velocity magnitude")
plt.legend()
plt.grid(True)

# pos compentents 
plt.subplot(2, 2, 3)
plt.plot(t, p_x, label="Horizontal displacement $p_x$")
plt.plot(t, p_y, label="Alltitudem $p_y$")
plt.xlabel("Time (s)")
plt.ylabel("Position (m)")
plt.title("Position against time")
plt.legend()
plt.grid(True)

#trajectory
plt.subplot(2, 2, 4)
plt.plot(p_x, p_y, label="Trajectory")
plt.xlabel("Horizontal displacement (m)")
plt.ylabel("Altitude (m)")
plt.title("Projectile path")
plt.legend()
plt.grid(True)

plt.tight_layout()   # stops subplots overlapping
plt.savefig("projectile_motion.png")
plt.show()





````
